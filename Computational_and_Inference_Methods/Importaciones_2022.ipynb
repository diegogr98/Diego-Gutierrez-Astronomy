```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
ruta_archivo = '/content/crecimiento_importaciones_2022.csv'  # Cambia por la ruta de tu archivo
df = pd.read_csv(ruta_archivo)
df
```

```python
df.info()
```

```python
# Identificar las columnas numéricas y convertirlas
columnas = [
    'Trade Value',
    'Trade Value Growth',
    'Trade Value Growth Value',
    'Trade Value Previous 1',
    'Trade Value Previous 3',
    'Trade Value Previous 5',
    'Trade Value Growth 1',
    'Trade Value Growth Value 1',
    'Trade Value Growth 3',
    'Trade Value Growth Value 3',
    'Trade Value Growth 5',
    'Trade Value Growth Value 5'
]  # Cambia por los nombres de tus columnas numéricas

def convertir_a_float(valor):
    if pd.isna(valor):  # Manejar valores NaN
        return valor
    try:
        # Reemplazar solo los separadores correctos: puntos por nada y comas por punto decimal
        valor = valor.replace('.', '').replace(',', '.')
        return float(valor)
    except AttributeError:
        # Si el valor no es una cadena (ya es NaN o un número), se devuelve sin cambios
        return valor
```

```python
for columna in columnas:
    df[columna] = df[columna].apply(convertir_a_float)
df
```

```python
# Encuentra el valor máximo de la columna "Trade Value" que equivale al avalor mas negociado
max_trade_value = df['Trade Value'].max()

# Encuentra la fila con el valor máximo de la columna "Trade Value"
max_trade_value_row = df.loc[df['Trade Value'] == max_trade_value]

# Imprime los valores de las columnas "HS2" y "HS4" de la fila con el valor máximo de la columna "Trade Value"
print("El valor máximo de la columna 'Trade Value' es:", max_trade_value)
print("HS2:", max_trade_value_row['HS2'].values[0])
print("HS2 ID:", max_trade_value_row['HS2 ID'].values[0])
print("HS4:", max_trade_value_row['HS4'].values[0])
print("HS4 ID:", max_trade_value_row['HS4 ID'].values[0])
```

```python
# Agrupar por 'HS2 ID' y 'HS2', y sumar los valores de 'Trade Value'
result = df.groupby(['HS2 ID', 'HS2'])['Trade Value'].sum().reset_index()

# Configurar Pandas para mostrar números de forma más legible (sin notación científica)
pd.set_option('display.float_format', '{:,.2f}'.format)

# Dividir los resultados en grupos de 6 filas
def dividir_en_grupos(dataframe, tamanio_grupo):
    return [dataframe.iloc[i:i + tamanio_grupo] for i in range(0, len(dataframe), tamanio_grupo)]

# Dividir el DataFrame agrupado en grupos de 6
grupos = dividir_en_grupos(result, 6)

# Mostrar los grupos
for i, grupo in enumerate(grupos, start=1):
    print(f"Grupo {i}:\n{grupo}\n")
```

```python
# Agrupar por 'HS4' y sumar los valores de 'Trade Value'
resultadomenor = df.groupby(['HS4'])['Trade Value'].sum().reset_index()

# Ordenar los resultados por la columna 'Trade Value' en orden ascendente y seleccionar los 10 primeros
bottom_10 = resultadomenor.sort_values(by='Trade Value', ascending=True).head(10)

# Configurar Pandas para mostrar números de forma más legible (sin notación científica)
pd.set_option('display.float_format', '{:,.2f}'.format)

# Mostrar los 10 productos menos negociados
print("Los 10 productos menos negociados (por Trade Value):\n")
print(bottom_10)

# Crear un gráfico tipo torta
plt.figure(figsize=(10, 6))
plt.pie(bottom_10['Trade Value'], labels=bottom_10['HS4'], autopct='%1.1f%%', startangle=140)
plt.title('Distribución de los Productos Menos Negociados por Trade Value')
plt.axis('equal')  # Para que el gráfico sea un círculo
plt.show()
```

```python
# Agrupar por 'HS4' y sumar los valores de 'Trade Value'
Top10 = df.groupby(['HS4'])['Trade Value'].sum().reset_index()

# Ordenar los resultados por la columna 'Trade Value' en orden descendente y seleccionar los 10 primeros
top_10 = Top10.sort_values(by='Trade Value', ascending=False).head(10)

# Configurar Pandas para mostrar números de forma más legible (sin notación científica)
pd.set_option('display.float_format', '{:,.2f}'.format)

# Mostrar los 10 productos más negociados
print("Los 10 productos más negociados (por Trade Value):\n")
print(top_10)

# Crear un gráfico tipo torta
plt.figure(figsize=(10, 6))
plt.pie(top_10['Trade Value'], labels=top_10['HS4'], autopct='%1.1f%%', startangle=140)
plt.title('Distribución de los Productos Más Negociados por Trade Value')
plt.axis('equal')  # Para que el gráfico sea un círculo
plt.show()
```

```python
# Seleccionar los diez productos con mayor crecimiento
bottom_growth = df.nlargest(10, "Trade Value Growth")

# Imprimir los productos con mayor crecimiento
print("Los diez productos con mayor crecimiento:\n")
print(bottom_growth[['HS4', 'Trade Value Growth']])

# Crear gráfico de barras
plt.figure(figsize=(12,6))

# Crear dos conjuntos de datos: uno para el crecimiento y otro para el valor absoluto
bar_width = 0.35
index = range(len(bottom_growth))

# Gráfico para Trade Value Growth
plt.bar(index,
        bottom_growth['Trade Value Growth'],
        bar_width, label='Crecimiento (%)',
        color='salmon')

# Configurar etiquetas y título
plt.yscale('log')
plt.xlabel('Productos')
plt.ylabel('Incremento porcentual')
plt.title('productos con mayor crecimiento de importaciones')
plt.xticks([i + bar_width / 2 for i in index], bottom_growth['HS4'], rotation=45)
plt.legend()

# Mostrar la gráfica
plt.tight_layout()
plt.show()
```
